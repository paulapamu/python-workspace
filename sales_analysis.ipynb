# Sales Analysis Project - This notebook is a project to become familiar with all the different libraries in python. 

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import scipy
```

##step 1 - Creating our Sales Data

```python
data = {
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "sales": [15000, 18000, 14000, 22000, 19000, 25000],
    "costs": [8000, 9000, 7500, 11000, 9500, 12000]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.describe())
```

```python
df["profit"]=df["sales"]-df["costs"]
print(df)
```

```python
df["goal"]=[10000, 65000, 2332, 3533,80000, 900000]
```

```python
print(df)
```

## Step 2 - Visualising our Sales Data with Matplotlib

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8), facecolor="white")

# Top panel - Sales and Costs
ax1.plot(df["month"], df["sales"], color="#2196F3", linewidth=1.5, label="Sales")
ax1.plot(df["month"], df["costs"], color="#2196F3", linewidth=1.5, linestyle="--", label="Costs")
ax1.set_ylabel("Sales (£)", fontsize=12)
ax1.spines["top"].set_visible(False)
ax1.spines["right"].set_visible(False)

# Bottom panel - Profit
ax2.plot(df["month"], df["profit"], color="#F44336", linewidth=1.5, label="Profit")
ax2.set_ylabel("Profit (£)", fontsize=12)
ax2.set_xlabel("Month", fontsize=12)
ax2.spines["top"].set_visible(False)
ax2.spines["right"].set_visible(False)

fig.legend(loc="lower center", ncol=2, frameon=False, fontsize=11)
plt.tight_layout()
plt.show()
```

## Step 3 - Calculations with NumPy

```python
avg_sales = np.mean(df["sales"])
avg_costs = np.mean(df["costs"])
avg_profit = np.mean(df["profit"])

print("Average sales: £" + str(avg_sales))
print("Average costs: £" + str(avg_costs))
print("Average profit: £" + str(avg_profit))
```

## Step 4 - Trend Analysis with SciPy

```python
from scipy import stats

months_numeric = [1, 2, 3, 4, 5, 6]

slope, intercept, r_value, p_value, std_err = stats.linregress(months_numeric, df["sales"])

print("Slope: " + str(slope))
print("R squared: " + str(r_value**2))

if p_value < 0.05:
    print("Sales are genuinely increasing over time!")
else:
    print("No clear trend in sales")
```
